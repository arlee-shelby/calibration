```python
def wave_fit(wave, func, pretrig=2000,p0=None):
    xdata = np.arange(0,len(wave))
    prewave = wave[0:int(pretrig)]
    base = np.average(prewave)
    ydata = wave - base
    if p0 == None:
        popt, pcov = curve_fit(func, xdata[3300:3700], ydata[3300:3700],maxfev=4000)
        res = sum((ydata[int(popt[4])-7:int(popt[4])+7].compute()- wave_func(xdata[int(popt[4])-7:int(popt[4])+7],*popt))**2)
    else:
        try:
            if type(ydata)==np.ndarray:
                popt, pcov = curve_fit(func, xdata, ydata, p0)
                res = sum((ydata[int(popt[4])-7:int(popt[4])+7]- wave_func(xdata[int(popt[4])-7:int(popt[4])+7],*popt))**2)             
            else:

                popt, pcov = curve_fit(func, xdata[2500:4500], ydata[2500:4500], p0)
                res = sum((ydata[int(popt[4])-7:int(popt[4])+7].compute()- wave_func(xdata[int(popt[4])-7:int(popt[4])+7],*popt))**2)

        except Exception as e:
            
            return 0,0,0
    
    return func(xdata, *popt) + base, popt, res

def wave_func(x, a,b,c,d,e,f):
    return a + b*x + c*x**2.0 + d*np.exp(-(x-e)/1250.0) / (1.0 + np.exp(-(x-e)/f))

def risetime(wave,fit_mid,low=0.1, high=0.9,pretrigger=[0,1000], cut=0,max_delay=0, max_length=50):
    try:
#         shift = np.mean(wave[pretrigger[0]:pretrigger[1]])
#         wave = wave - shift
        if max_delay == 0:
            shift = np.mean(wave[pretrigger[0]:pretrigger[1]])
            wave = wave - shift
            maxval = np.max(wave[int(fit_mid)-15:int(fit_mid)+15])
        else:
            fit_mid = int(fit_mid)-2500
            shift = np.mean(wave[fit_mid-200:fit_mid-100])
            wave = wave - shift
            maxval = max(wave)

        wave = wave[cut:]
        xs = np.arange(len(wave))*4
        upperCross = maxval * high
        lowerCross = maxval * low
        
        upCrossPoint = np.argwhere(wave[int(fit_mid):]>upperCross)[0][0]+int(fit_mid)
        lowCrossPoint = np.argwhere(wave[:int(fit_mid)]<lowerCross)[-1][0]+1
        
        upper = np.interp(upperCross, wave[upCrossPoint-1:upCrossPoint+1], xs[upCrossPoint-1:upCrossPoint+1])
        lower = np.interp(lowerCross, wave[lowCrossPoint-1:lowCrossPoint+1], xs[lowCrossPoint-1:lowCrossPoint+1])
        
        return upper-lower, [lowCrossPoint, upCrossPoint, maxval+shift]
    
    except Exception as e:
#         plt.plot(wave,label = 'exception wave')
#         plt.legend()
#         plt.show()
        print(e)
        return 0,[0,0,0]
```

```python
def apply_mask(props, applied_mask):
    masked_props = []
    for prop in props:
        masked_props.append(prop[applied_mask])
    return masked_props

def apply_mask_to_set(masks, applied_mask):
    masked_masks = []
    for var_masks in masks:
        masked_masks.append(apply_mask(var_masks,applied_mask))
    return masked_masks

def get_cut_masks(var,cuts):
    masks = []
    for cut in cuts:
        masks.append(var<cut)
    return masks

def get_mask_set(var_list,var_cuts):
    mask_set = []
    for i in range(len(var_list)):
        mask_set.append(get_cut_masks(var_list[i],var_cuts[i]))
    return mask_set

def get_masked_risetime_props(masks, num_waves_list, waveforms, waveform_risetimes):
    rise_app = []
    rise_app_num = []
    rise_app_dis = []
    for j in range(len(masks)):
        rise_av = []
        num = []
        total_singles = []
        
        total  = np.array(waveforms)[masks[j]]
        singles = np.array(waveform_risetimes)[masks[j]]
        
        for i in num_waves_list:
#             print(i,len(total))
            if i>len(total):
                pass
            else:
                random_pairs = random.sample(list(zip(total,singles)),i)
                rand_single_rise = [x[1] for x in random_pairs]
                rand_waves = [x[0] for x in random_pairs]
#                 print(len(rand_waves))
#                 print(i)
#                 if i==50:
#                     print(rand_waves)
                n = len(rand_waves)
#                 print(sum(rand_waves))
                average = sum(rand_waves)/len(rand_waves)
#                 plt.plot(average)
#                 plt.show()
                func, par, res = wave_fit(average, wave_func, p0=[5,5e-3,1e-7,50,3500,0.7])
                rise, vals = risetime(average, par[4],low=0.1, high=0.9)
                rise_av.append(rise)
                total_singles.append(rand_single_rise)
                num.append(n)
        
        rise_app.append(rise_av)
        rise_app_dis.append(total_singles)
        rise_app_num.append(num)
        
    return rise_app, rise_app_num, rise_app_dis

def get_masked_risetime_set(mask_set,num_waves_list,waveforms,waveform_risetimes):
    props = []
    for masks in mask_set:
        average_rise, num_waves, rise_dis = get_masked_risetime_props(masks,num_waves_list,waveforms,waveform_risetimes)
        props.append([average_rise, num_waves, rise_dis])
    return props
```
