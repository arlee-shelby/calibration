```python
def wave_fit(wave, func, pretrig=2000,p0=None):
    xdata = np.arange(0,len(wave))
    prewave = wave[0:int(pretrig)]
    base = np.average(prewave)
    ydata = wave - base
    if p0 == None:
        popt, pcov = curve_fit(func, xdata[3300:3700], ydata[3300:3700],maxfev=4000)
        res = sum((ydata[int(popt[4])-7:int(popt[4])+7].compute()- wave_func(xdata[int(popt[4])-7:int(popt[4])+7],*popt))**2)/len(ydata[int(popt[4])-7:int(popt[4])+7].compute())
    else:
        try:
            if type(ydata)==np.ndarray:
                popt, pcov = curve_fit(func, xdata, ydata, p0)
                res = sum((ydata[int(popt[4])-7:int(popt[4])+7]- wave_func(xdata[int(popt[4])-7:int(popt[4])+7],*popt))**2)/len(ydata[int(popt[4])-7:int(popt[4])+7])         
            else:

                popt, pcov = curve_fit(func, xdata[2500:4500], ydata[2500:4500], p0)
                res = sum((ydata[int(popt[4])-7:int(popt[4])+7].compute()- wave_func(xdata[int(popt[4])-7:int(popt[4])+7],*popt))**2)/len(ydata[int(popt[4])-7:int(popt[4])+7].compute())

        except Exception as e:
            
            return 0,0,0
    
    return func(xdata, *popt) + base, popt, res

def wave_func(x, a,b,c,d,e,f):
    return a + b*x + c*x**2.0 + d*np.exp(-(x-e)/1250.0) / (1.0 + np.exp(-(x-e)/f))

def risetime(wave,fit_mid,low=0.1, high=0.9,pretrigger=[0,1000], cut=0,max_delay=0, max_length=50):
    try:
#         shift = np.mean(wave[pretrigger[0]:pretrigger[1]])
#         wave = wave - shift
        if max_delay == 0:
            shift = np.mean(wave[pretrigger[0]:pretrigger[1]])
            wave = wave - shift
            maxval = np.max(wave[int(fit_mid)-15:int(fit_mid)+15])
        else:
            fit_mid = int(fit_mid)-2500
            shift = np.mean(wave[fit_mid-200:fit_mid-100])
            wave = wave - shift
            maxval = max(wave)

        wave = wave[cut:]
        xs = np.arange(len(wave))*4
        upperCross = maxval * high
        lowerCross = maxval * low
        
        upCrossPoint = np.argwhere(wave[int(fit_mid):]>upperCross)[0][0]+int(fit_mid)
        lowCrossPoint = np.argwhere(wave[:int(fit_mid)]<lowerCross)[-1][0]+1
        
        upper = np.interp(upperCross, wave[upCrossPoint-1:upCrossPoint+1], xs[upCrossPoint-1:upCrossPoint+1])
        lower = np.interp(lowerCross, wave[lowCrossPoint-1:lowCrossPoint+1], xs[lowCrossPoint-1:lowCrossPoint+1])
        
        return upper-lower, [lowCrossPoint, upCrossPoint, maxval+shift]
    
    except Exception as e:
#         plt.plot(wave,label = 'exception wave')
#         plt.legend()
#         plt.show()
        print(e)
        return 0,[0,0,0]
```

```python
def apply_mask(props, applied_mask):
    masked_props = []
    for prop in props:
        masked_props.append(prop[applied_mask])
    return masked_props

def apply_mask_to_set(masks, applied_mask):
    masked_masks = []
    for var_masks in masks:
        masked_masks.append(apply_mask(var_masks,applied_mask))
    return masked_masks

def get_cut_masks(var,cuts):
    masks = []
    for cut in cuts:
        masks.append(var<cut)
    return masks

def get_mask_set(var_list,var_cuts):
    mask_set = []
    for i in range(len(var_list)):
        mask_set.append(get_cut_masks(var_list[i],var_cuts[i]))
    return mask_set

def get_masked_risetime_props(masks, num_waves_list, waveforms, waveform_risetimes):
    rise_app = []
    rise_app_num = []
    rise_app_dis = []
    for j in range(len(masks)):
        rise_av = []
        num = []
        total_singles = []
        for i in num_waves_list:
#             if direction == 'first':
            total  = np.array(waveforms)[masks[j]][:i]
            n = len(total)
            singles = np.array(waveform_risetimes)[masks[j]][:i]
#             else:
#                 total  = np.array(waveforms)[masks[j]][-i:]
#                 n = len(total)
#                 singles = np.array(waveform_risetimes)[masks[j]][-i:]
            average = sum(total)/len(total)
            func, par, res = wave_fit(average, wave_func, p0=[5,5e-3,1e-7,50,3500,0.7])
            rise, vals = risetime(average, par[4],low=0.1, high=0.9)
            rise_av.append(rise)
            total_singles.append(singles)
            num.append(n)
        
        rise_app.append(rise_av)
        rise_app_dis.append(total_singles)
        rise_app_num.append(num)
        
    return rise_app, rise_app_num, rise_app_dis

def get_masked_risetime_set(mask_set,num_waves_list,waveforms,waveform_risetimes):
    props = []
    for masks in mask_set:
        average_rise, num_waves, rise_dis = get_masked_risetime_props(masks,num_waves_list,waveforms,waveform_risetimes)
        props.append([average_rise, num_waves, rise_dis])
    return props
```

```python
def get_waves(run, pixel):
    
    run.singleWaves().resetCuts()
    print(len(run.singleWaves().waves()))
    run.singleWaves().defineCut('source', '=', 0)
    print(len(run.singleWaves().waves()))
    run.singleWaves().defineCut('pixel', '=', pixel)

    results = run.singleWaves().determineEnergyTiming(method='trap', params=[1250, 50, 1250])

    results.resetCuts()
    print(len(run.singleWaves().waves()))

    results.defineCut('energy', 'between',35,40)
    cut = results.returnCut()
    run.singleWaves().defineCut('custom', cut)
    waves = run.singleWaves().waves()
    print(len(waves))

    before = 0
    after = 0
    num = 0
    rise_single = []
    res_single = []
    var2_single = []
    var3_single = []
    waves_single = []
    par_single = []
    func_single = []
    for i in range(len(waves)):
        ydata = waves[i]
        func, par, res = wave_fit(waves[i],wave_func,p0=[5,5e-3,1e-7,50,3500,0.7])

        if type(par)==int:
            pass

        elif (np.abs(par[5])/par[5] != np.abs(par[3])/par[3]):
            pass

        elif np.abs(par[5])/par[5] == np.abs(par[3])/par[3]:
            rise, vals = risetime(func[2500:4500].compute(), par[4],low=0.1, high=0.9,max_delay=2500)

            var3 = sum((ydata[int(par[4])+7:int(par[4])+200].compute() - np.mean(ydata[int(par[4])+7:int(par[4])+200].compute()))**2)/len(ydata[int(par[4])+7:int(par[4])+200].compute())
            var2 = sum((ydata[int(par[4])-200:int(par[4])-7].compute() - np.mean(ydata[int(par[4])-200:int(par[4])-7].compute()))**2)/len(ydata[int(par[4])-200:int(par[4])-7].compute())
            res = sum((ydata[vals[0]+2500:vals[1]+2500].compute() - func[vals[0]+2500:vals[1]+2500].compute())**2)/len(ydata[vals[0]+2500:vals[1]+2500].compute())

            par_single.append(par)
            rise_single.append(rise)
            res_single.append(res)
            var2_single.append(var2)
            var3_single.append(var3)

            start = int(par[4])-2900
            end = int(par[4]) + 3000

            wave_before = np.array(waves[i][start:int(par[4])].compute())
            wave_after = np.array(waves[i][int(par[4]):end].compute())

            func_before = np.array(func[start:int(par[4])].compute())
            func_after = np.array(func[int(par[4]):end].compute())

            wv = np.concatenate((wave_before, wave_after))
            fn = np.concatenate((func_before, func_after))
            waves_single.append(wv)
            func_single.append(fn)
            
    waves_and_pars = [np.array(waves_single), np.array(func_single), np.array(par_single), np.array(rise_single),
                      np.array(res_single), np.array(var2_single), np.array(var3_single)]
    return waves_and_pars



def get_average(waves_and_pars,runNumber,pixel,bias):
    var_list = [waves_and_pars[4],waves_and_pars[5],waves_and_pars[6]]
    mask_set = get_mask_set(var_list,var_cuts)
    
    num_waves = [100,200,300,500,800,1000]
    av_result_from_masks = get_masked_risetime_set(mask_set,num_waves,waves_and_pars[0],waves_and_pars[3])
    
    var3_cut_mask = mask_set[2][8]
    print(var2_cuts[8])
    mask_set_var3_cut = apply_mask_to_set(mask_set,var3_cut_mask)
    waves_and_pars_var3_cut = apply_mask(waves_and_pars,var3_cut_mask)
    waves_var3_cut = waves_and_pars_var3_cut[0]
    rise_var3_cut = waves_and_pars_var3_cut[3]
    total_num_waves_var3_cut = len(waves_var3_cut)
    num_waves_var3_cut = [100,200,300,500,800,1000]
    print(total_num_waves_var3_cut)

    av_result_from_masks_var3_cut = get_masked_risetime_set(mask_set_var3_cut,num_waves_var3_cut,
                                                                waves_var3_cut,rise_var3_cut)

    var3_and_var2_cut_mask = mask_set_var3_cut[1][8]
    print(var3_cuts[8])
    mask_set_var3_and_var2_cut = apply_mask_to_set(mask_set_var3_cut,var3_and_var2_cut_mask)
    waves_and_pars_var3_and_var2_cut = apply_mask(waves_and_pars_var3_cut,var3_and_var2_cut_mask)
    waves_var3_and_var2_cut = waves_and_pars_var3_and_var2_cut[0]
    rise_var3_and_var2_cut = waves_and_pars_var3_and_var2_cut[3]
    total_num_waves_var3_and_var2_cut = len(waves_var3_and_var2_cut)
    num_waves_var3_and_var2_cut = [100,200,300,500,800]
    print(total_num_waves_var3_and_var2_cut)

    av_result_from_masks_var3_and_var2_cut = get_masked_risetime_set(mask_set_var3_and_var2_cut,num_waves_var3_and_var2_cut,
                                                                         waves_var3_and_var2_cut,rise_var3_and_var2_cut)

    if total_num_waves_var3_and_var2_cut>1000:
        waves = random.sample(list(waves_var3_and_var2_cut),1000)
    else:
        waves = waves_var3_and_var2_cut

    n = len(waves)
    average = sum(waves)/n
    func, par, res = wave_fit(average, wave_func, p0=[5,5e-3,1e-7,50,3500,0.7])
    rise, vals = risetime(average, par[4],low=0.1, high=0.9)
    plt.plot(average)
    plt.plot(func,label = 'risetime:%.2f'%rise)
    plt.legend()
    plt.title('run %d, -%d V, pixel %d'%(runNumber,bias,pixel))
    plt.show()
    np.savetxt('protons_pixel%d_-%dV.csv'%(pixel,bias), average, delimiter=',')
    return average, func, rise
```
